# Cleaning Text

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\S+', ' ', text)
    text = re.sub(r'[\n\t\r]+', ' ', text)
    text = re.sub(r'[^a-zA-Z0-9+#.\- ]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df['Resume']=df['Resume'].apply(clean_text)
```

```python
def remove_header(text):
    text = re.sub(r'^.*?\s-\s', '', text, count=1)
    return text.strip()
```

```python
df['Resume']=df['Resume'].apply(remove_header)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
df['label']=le.fit_transform(df['Category'])
```

# Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
x = df['Resume']
y = df['label']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=809)
```

```python
text=[sentence.split() for sentence in x_train]
print(text[0])
```

```python
from gensim.models import Word2Vec

w2v=Word2Vec(min_count=2, window=10, epochs=20, workers=4, vector_size=100)
```

```python
w2v.build_vocab(text)
print(w2v.wv)
```

```python
w2v.train(text, total_examples=w2v.corpus_count, epochs=10)
```

```python
w2v.wv.similarity('python', 'machine')
```

# Tokenization

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tk=Tokenizer(num_words=1000, oov_token="<OOV>")
tk.fit_on_texts(x_train)
```

```python
x_train_seq=tk.texts_to_sequences(x_train)
x_test_seq=tk.texts_to_sequences(x_test)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

max_len=500
x_train_pad=pad_sequences(x_train_seq, maxlen=max_len, padding='pre', truncating='post')
x_test_pad=pad_sequences(x_test_seq, maxlen=max_len, padding='pre', truncating='post')
```

```python
x_train_pad[10]
```

# Creating Embedding Matrix

```python
embedding_dim=100
vocab_size=20000

embedding_matrix=np.zeros((vocab_size, embedding_dim))
word_index=tk.word_index

for word, index in word_index.items():
    if index>=vocab_size:
        continue
    if word in w2v.wv:
        embedding_matrix[index]=(w2v.wv[word])
```

```python
embedding_matrix
```

# Model Training

```python
from keras.models import Sequential
from keras.layers import Input, Dense, Embedding, Bidirectional, LSTM, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
embedding=Embedding(input_dim=20000, output_dim=100, trainable=True, weights=[embedding_matrix])
```

```python
model=Sequential()

model.add(Input(shape=(500,)))
model.add(embedding)

model.add(Bidirectional(LSTM(128)))
model.add(Dropout(0.3))

model.add(Dense(32, activation='relu'))
model.add(Dropout(0.3))

model.add(Dense(10, activation='softmax'))
model.summary()
```

```python
early_stopping=EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['Accuracy'])
```

```python
model.fit(x_train_pad, y_train, epochs=100, batch_size=256, validation_split=0.2, callbacks=[early_stopping])
```

```python
def predict(resume):
    cleaned = clean_text(resume)
    cleaned = remove_header(cleaned)

    sequence = tk.texts_to_sequences([cleaned])

    padded = pad_sequences(sequence,maxlen=500,padding='pre',truncating='post')

    probabilities = model.predict(padded, verbose=0)[0]
    top_5_indices = np.argsort(probabilities)[::-1][:5]

    results = []

    print("Top 5 Recommended Roles:\n")

    for rank, index in enumerate(top_5_indices, start=1):

        role = le.inverse_transform([index])[0]
        probability = probabilities[index]

        print(f"{rank}. {role}: {probability * 100:.2f}%")

        results.append({"role": role,"probability": probability})
```
